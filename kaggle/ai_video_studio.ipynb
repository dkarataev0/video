### Шаг 1. Скачать программу

```python
REPO = "github.com/dkarataev0/video.git"
BRANCH = "claude/face-swap-voice-clone-yijuew"   # после слияния в main можно указать "main"

import os, subprocess
token = ""
try:  # для приватного репозитория: Add-ons → Secrets → GITHUB_TOKEN
    from kaggle_secrets import UserSecretsClient
    token = UserSecretsClient().get_secret("GITHUB_TOKEN")
except Exception:
    pass
url = f"https://{token + '@' if token else ''}{REPO}"
subprocess.run(["rm", "-rf", "/kaggle/working/video"])
r = subprocess.run(["git", "clone", "-q", "--depth", "1", "-b", BRANCH, url, "/kaggle/working/video"])
print("✅ Готово" if r.returncode == 0 else "❌ Не удалось скачать: проверьте, что репозиторий публичный или задан секрет GITHUB_TOKEN")
```
